```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

data = {
    'Student': ['Amit','Neha','Rahul','Priya','Karan','Anjali','Ravi','Sneha'],
    'Marks': [85, 92, 78, 88, 88, 95, 70, 90],
    'Attendance': [90, 95, 80, 85, 88, 98, 75, 92],
    'Study_Hours': [4, 6, 3, 5, 4, 7, 2, 6],
}

df = pd.DataFrame(data)
print('DATASET')
print(df)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df['Student'], df['Marks'], marker='o', linewidth=2)
plt.title('Student-wise Marks')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.bar(df['Student'], df['Marks'])
plt.title('Student Performance')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Study_Hours'], df['Marks'], marker='o', s=100)
plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Marks'], bins=5)
plt.title('Distribution of Marks')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df[['Marks', 'Attendance', 'Study_Hours']])
plt.title('Box Plot of Student Performance')
plt.ylabel('Values')
plt.tight_layout()
plt.show()
```

```python
correlation = df[['Marks', 'Attendance', 'Study_Hours']].corr()
print('\nCORRELATION MATRIX')
print(correlation)
```

```python
plt.figure(figsize=(7, 5))
sns.heatmap(correlation, annot=True)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Study_Hours'], df['Marks'], marker='o', s=120)
plt.title('Relationship Between Study Hours and Marks', fontsize=14)
plt.xlabel('Study Hours', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.bar(df['Student'], df['Marks'])
plt.title('Student Marks')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('student_marks.png')
plt.show()

print('\nPlot saved successfully as student_marks.png')
```

```python
plt.figure(figsize=(8, 5))
plt.barh(df['Student'], df['Marks'])
plt.title('Student Marks - Horizontal Bar Chart')
plt.xlabel('Marks')
plt.ylabel('Student')
plt.grid(axis='x')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Attendance'], df['Marks'], s=120)
plt.title('Attendance vs Marks')
plt.xlabel('Attendance (%)')
plt.ylabel('Marks')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Attendance'], bins=5)
plt.title('Distribution of Attendance')
plt.xlabel('Attendance (%)')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(y=df['Marks'])
plt.title('Box Plot of Marks')
plt.ylabel('Marks')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    df['Study_Hours'],
    labels=df['Student'],
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Percentage Distribution of Study Hours')
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(df[['Marks', 'Attendance', 'Study_Hours']])
plt.suptitle('Pairwise Relationship Between Variables', y=1.02)
plt.show()
```
