```python
import pandas as pd

data = {
    'Student_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Name': ['Amit', 'Neha', 'Rahul', 'Priya', 'Karan', 'Anjali', 'Ravi', 'Sneha'],
    'Department': ['CSE', 'CSE', 'ECE', 'ECE', 'CSE', 'IT', 'IT', 'CSE'],
    'Gender': ['M', 'F', 'M', 'F', 'M', 'F', 'M', 'F'],
    'Marks': [85, 92, 78, None, 88, 95, 70, 90],
    'Attendance': [90, 95, 80, 85, None, 98, 75, 92],
    'Age': [20, 21, 20, 22, 21, 20, 22, 21],
    'Remarks': ['Good','Excellent','Good','Average','Good','Excellent','Average','Excellent',],
}

df = pd.DataFrame(data)
print('\nORIGINAL DATASET\n')
print(df)
```

```python
print('DATASET INFORMATION')
print(df.info())
```

```python
print('MISSING VALUES')
print(df.isnull().sum())
```

```python
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mean())
print('DATA AFTER HANDLING MISSING VALUES\n')
print(df)
```

```python
df = df.drop('Remarks', axis=1)
print('DATA AFTER REMOVING UNNECESSARY COLUMN\n')
print(df)
```

```python
print('MISSING VALUES AFTER CLEANING\n')
print(df.isnull().sum())
```

```python
print('DEPARTMENT-WISE AVERAGE MARKS\n')
department_average = df.groupby('Department')['Marks'].mean()
print(department_average)
```

```python
print('DEPARTMENT-WISE SUMMARY STATISTICS\n')
summary = df.groupby('Department')['Marks'].agg(['mean', 'sum', 'min', 'max', 'count'])
print(summary)
```

```python
print('DEPARTMENT-WISE AVERAGE ATTENDANCE\n')
attendance = df.groupby('Department')['Attendance'].mean()
print(attendance)
```

```python
print("GENDER-WISE AVERAGE MARKS") 
gender_average = df.groupby('Gender')['Marks'].mean() 
print(gender_average) 
```

```python
print('OVERALL SUMMARY STATISTICS\n')
print(df.describe())
```
