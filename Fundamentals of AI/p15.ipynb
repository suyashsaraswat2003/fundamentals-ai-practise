```python
def admission_chatbot():
    print("SRM Admission Enquiry Chatbot")
    print("Type 'bye' to exit.")

    while True:
        user = input("\nYou: ").lower().strip()

        if user == "bye":
            print("Bot: Thank you for contacting SRM. Goodbye!")
            break

        elif "hello" in user or "hi" in user:
            print("Bot: Hello! Welcome to SRM admission enquiry.")

        elif "course" in user or "programme" in user:
            print("Bot: SRM offers programmes in Engineering, "
                  "Management, Science, and Humanities.")

        elif "eligibility" in user or "qualification" in user:
            print("Bot: Eligibility depends on the programme. "
                  "Please check the official SRM admissions website.")

        elif "fee" in user or "fees" in user:
            print("Bot: Fees vary by programme and campus. "
                  "Please consult the official fee structure.")

        elif "admission" in user or "apply" in user:
            print("Bot: Visit the official SRM website for "
                  "application details and admission updates.")

        elif "exam" in user or "entrance" in user:
            print("Bot: Entrance examination requirements depend "
                  "on the programme.")

        elif "contact" in user:
            print("Bot: Please visit the official SRM website "
                  "for verified contact information.")

        else:
            print("Bot: Sorry, I could not understand that. "
                  "Please ask about courses, fees, eligibility, "
                  "or admissions.")

admission_chatbot()
```
