```python
import pandas as pd

data = {
    'Name': ['Amit','Neha','Rahul','Priya','Karan','Anjali','Ravi','Sneha','Vikas','Pooja','Arjun','Simran',],
    'Department': ['CSE','CSE','ECE','ECE','CSE','IT','IT','CSE','ECE','IT','CSE','IT',],
    'Marks': [85, 92, 78, 88, 75, 95, 70, 90, 82, 89, 96, 73],
    'Age': [20, 21, 20, 22, 21, 20, 22, 21, 20, 21, 22, 20],
}

df = pd.DataFrame(data)

df.to_csv('s1.csv', index=False)
print('CSV file created successfully.')
```

```python
df = pd.read_csv('s1.csv')
print('\nCOMPLETE DATASET')
print(df)
```

```python
print('\nFIRST 10 ROWS')
print(df.head(10))
```

```python
print('\nDATA SORTED BY MARKS')
sorted_df = df.sort_values(by='Marks', ascending=False)
print(sorted_df)
```

```python
print('\nSTUDENTS WITH MARKS > 80')
filtered_df = df[df['Marks'] > 80]
print(filtered_df)
```

```python
print('\nCSE STUDENTS')
cse_students = df[df['Department'] == 'CSE']
print(cse_students)
```

```python
print('\nMARKS > 80 AND AGE < 22')
result = df[(df['Marks'] > 80) & (df['Age'] < 22)]
print(result)
```
