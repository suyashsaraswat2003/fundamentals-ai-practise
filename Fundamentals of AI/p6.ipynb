```python
import random

class VacuumEnvironment:

    def __init__(self, status=None, start_location=None):
        self.status = status.copy() if status else {
            'A': random.choice(['Dirty', 'Clean']), 
            'B': random.choice(['Dirty', 'Clean'])}
        self.agent_location = start_location or random.choice(['A', 'B'])
        self.performance = 0
        self.steps = 0

    def percept(self):
        return self.agent_location, self.status[self.agent_location]

    def execute(self, action):
        self.steps += 1
        self.performance -= 1

        if action == "Suck":
            if self.status[self.agent_location] == "Dirty":
                self.status[self.agent_location] = "Clean"
                self.performance += 10
        elif action == "Right" and self.agent_location == "A":
            self.agent_location = "B"
        elif action == "Left" and self.agent_location == "B":
            self.agent_location = "A"

    def is_clean(self):
        return self.status['A'] == 'Clean' and self.status['B'] == 'Clean'


def simple_reflex_agent(percept):
    location, status = percept

    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Right"
    else:
        return "Left"


class ModelBasedReflexAgent:

    def __init__(self):
        self.model = {'A': None, 'B': None}

    def program(self, percept):
        location, status = percept
        self.model[location] = status

        if self.model['A'] == 'Clean' and self.model['B'] == 'Clean':
            return "NoOp"
        if status == "Dirty":
            return "Suck"
        elif location == "A":
            return "Right"
        else:
            return "Left"


def run_agent(agent, environment, max_steps=20):

    print("Initial Environment:", environment.status)
    print("Starting Location:", environment.agent_location)
    print("\nExecution:")

    for step in range(1, max_steps + 1):
        percept = environment.percept()
        action = agent(percept) if callable(agent) else agent.program(percept)

        print(f"Step {step} | Percept = {percept} | Action = {action} | Rooms = {environment.status} | Score = {environment.performance}")

        environment.execute(action)

        if action == "NoOp" or environment.is_clean():
            break

    print("\nFinal Environment:", environment.status)
    print("Final Location:", environment.agent_location)
    print("Total Steps:", environment.steps)
    print("Final Performance:", environment.performance)

    return environment.performance


if __name__ == "__main__":

    environment1 = VacuumEnvironment(status={'A': 'Dirty', 'B': 'Dirty'}, start_location='A')
    simple_score = run_agent(simple_reflex_agent, environment1, max_steps=10)

    environment2 = VacuumEnvironment(status={'A': 'Dirty', 'B': 'Dirty'}, start_location='A')
    model_agent = ModelBasedReflexAgent()
    model_score = run_agent(model_agent, environment2, max_steps=10)

    print("\nPerformance Comparison:")
    print("Simple Reflex Agent:", simple_score)
    print("Model-Based Reflex Agent:", model_score)
    print("\nInternal Model:", model_agent.model)

    print("\nSummary")
    print(""" 1. Vacuum Cleaner is the AI Agent. 
    2. Room A and Room B form the Environment. 
    3. The agent receives a Percept. 
    4. The agent selects an Action. 
    5. Suck cleans the current room. 
    6. Right moves from A to B. 
    7. Left moves from B to A. 
    8. Simple Reflex Agent uses the current percept. 
    9. Model-Based Reflex Agent maintains an internal model. 
    10. The goal is to make both rooms Clean. """)
```
