```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original Array:")
print(arr)

print("\nArray Shape:")
print(arr.shape)

print("\nArray Size:")
print(arr.size)

print("\nArray Data Type:")
print(arr.dtype)

print("\nArray Data Type:")
print(arr.dtype)

print("\nArray Data Type:")
print(arr.dtype)

print("\nAddition by 5:")
print(arr + 5)

print("\nSubtraction by 5:")
print(arr - 5)

print("\nMultiplication by 2:")
print(arr * 2)

print("\nDivision by 2:")
print(arr / 2)

print("\nSum of Array:")
print(np.sum(arr))

print("\nMean of Array:")
print(np.mean(arr))

print("\nMaximum Value:")
print(np.max(arr))

print("\nMinimum Value:")
print(np.min(arr))

print("\nStandard Deviation:")
print(np.std(arr))

print("\nFirst Element:")
print(arr[0])

print("\nLast Element:")
print(arr[-1])

print("\nFirst Three Elements:")
print(arr[:3])

print("\nElements from Index 2:")
print(arr[2:])
```
