```python
while True:
    print("\n  PYTHON BASICS & CONTROL STRUCTURES")
    print("--------------------------------------")
    print("1. Arithmetic Operations")
    print("2. Check Even or Odd")
    print("3. Check Positive, Negative or Zero")
    print("4. Find Largest of Three Numbers")
    print("5. Multiplication Table")
    print("6. Factorial of a Number")
    print("7. Sum of First N Natural Numbers")
    print("8. Display Even Numbers from 1 to N")
    print("9. Demonstrate Break and Continue")
    print("10. Exit")
    
    choice = int(input("\nEnter your choice (1-10): "))

    # 1. Arithmetic Operations 
    if choice == 1:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))
        print("\nArithmetic Operations")
        print("Addition =", a + b)
        print("Subtraction =", a - b)
        print("Multiplication =", a * b)
        if b != 0:
            print("Division =", a / b)
            print("Floor Division =", a // b)
            print("Modulus =", a % b)
        else:
            print("Division, Floor Division, and Modulus not possible.")
        print("Exponentiation =", a ** b)

    # 2. Even or Odd 
    elif choice == 2:
        num = int(input("Enter a number: "))
        if num % 2 == 0:
            print(num, "is Even.")
        else:
            print(num, "is Odd.")

    # 3. Positive, Negative or Zero 
    elif choice == 3:
        num = float(input("Enter a number: "))
        if num > 0:
            print("Positive Number")
        elif num < 0:
            print("Negative Number")
        else:
            print("Zero")

    # 4. Largest of Three Numbers 
    elif choice == 4:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))
        c = float(input("Enter third number: "))
        if a >= b and a >= c:
            print("Largest Number =", a)
        elif b >= a and b >= c:
            print("Largest Number =", b)
        else:
            print("Largest Number =", c)

    # 5. Multiplication Table 
    elif choice == 5:
        num = int(input("Enter a number: "))
        print("\nMultiplication Table")
        for i in range(1, 11):
            print(f"{num} x {i} = {num * i}")

    # 6. Factorial 
    elif choice == 6:
        num = int(input("Enter a number: "))
        if num < 0:
            print("Factorial does not exist for negative numbers.")
        else:
            fact = 1
            for i in range(1, num + 1):
                fact *= i
            print("Factorial =", fact)

    # 7. Sum of First N Natural Numbers 
    elif choice == 7:
        n = int(input("Enter N: "))
        total = 0
        for i in range(1, n + 1):
            total += i
        print("Sum =", total)

    # 8. Even Numbers 
    elif choice == 8:
        n = int(input("Enter N: "))
        print("Even Numbers:")
        for i in range(2, n + 1, 2):
            print(i, end=" ")
        print()

    # 9. Break and Continue 
    elif choice == 9:
        print("\nBreak Statement Example")
        for i in range(1, 11):
            if i == 6:
                break
            print(i, end=" ")
        print("\n\nContinue Statement Example")
        for i in range(1, 11):
            if i == 6:
                continue
            print(i, end=" ")
        print()

    # 10. Exit 
    elif choice == 10:
        print("\nThank you! Program Terminated Successfully.")
        break
    else:
        print("Invalid Choice! Please enter a number between 1 and 10.")
```
