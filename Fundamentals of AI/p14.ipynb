```python
application_complete = True
can_be_reviewed = True

if application_complete:
    can_be_reviewed = True

print("Propositional Logic:")
print("Application can be reviewed:", can_be_reviewed)

rules = {
    "fever": "possible_infection",
    "cough": "possible_respiratory_problem",
    "low_attendance": "attendance_shortage"
}

facts = ["fever", "cough"]
print("\nRule-Based Reasoning:")

for fact in facts:
    if fact in rules:
        print(fact, "->", rules[fact])

student = {
    "name": "Ravi",
    "programme": "B.Tech",
    "attendance": 85,
    "application_status": "Submitted"
}

print("\nStudent Frame:")
for key, value in student.items():
    print(key, ":", value)

semantic_network = [
    ("Ravi", "INSTANCE_OF", "Student"),
    ("Student", "IS_A", "Person"),
    ("Ravi", "HAS", "Application")
]

print("\nSemantic Network:")

for subject, relation, obj in semantic_network:
    print(subject, "--", relation, "-->", obj)

if student["application_status"] == "Submitted":
    print("\nInference: Application has been submitted.")
```
