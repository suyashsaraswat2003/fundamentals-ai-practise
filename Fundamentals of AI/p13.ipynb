```python
def unify(x, y, subs=None):
    if subs is None:
        subs = {}

    if x == y:
        return subs

    if isinstance(x, str) and x.startswith("?"):
        subs[x] = y
        return subs

    if isinstance(y, str) and y.startswith("?"):
        subs[y] = x
        return subs

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y) or x[0] != y[0]:
            return None

        for a, b in zip(x[1:], y[1:]):
            subs = unify(a, b, subs)
            if subs is None:
                return None

        return subs

    return None

def resolve(clause1, clause2):
    for literal1 in clause1:
        for literal2 in clause2:
            if literal1[0] != literal2[0] and literal1[1:] == literal2[1:]:
                new_clause = [ l for l in clause1 if l != literal1 ] + [ l for l in clause2 if l != literal2 ]
                return list(set(new_clause))
    return None

result = unify( ("Symptom", "?x", "fever"), ("Symptom", "Ravi", "fever") )
print("Unification:", result)
clause1 = ["P", "Q"]
clause2 = ["NOT P"]

def propositional_resolution(c1, c2):
    for literal in c1:
        opposite = ( literal[4:] if literal.startswith("NOT ") else "NOT " + literal )
        if opposite in c2:
            return list(set( [x for x in c1 if x != literal] + [x for x in c2 if x != opposite] ))
   
    return None

print("Resolvent:", propositional_resolution(clause1, clause2))
```
